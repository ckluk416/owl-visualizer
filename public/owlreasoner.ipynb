```python
from owlready2 import *
```

```python
onto_path.append(".")
onto = get_ontology("BallpointPenOntology.owl").load()
```

```python
# 2. Buka 'ruang lingkup' ontologi untuk menambahkan ABox
with onto:
    # Membuat instance dari kelas BicCristal
    pen_ku = onto.BicCristal("Pena_Bic_Cristal_Biru")

    # Menambahkan Object Property
    pen_ku.hasCharacteristic.append(onto.Disposable)

    # (hasMake ADALAH functional property, jadi langsung assign nilainya)
    pen_ku.hasMake = onto.Bic

    # (hasWidestDiameterInMillimetres ADALAH functional property, jangan pakai list [])
    pen_ku.hasWidestDiameterInMillimetres = 8.5

print("Sebelum reasoning: Ontologi berhasil dimuat dan ABox ditambahkan.")
```

```python
with onto:
    sync_reasoner()

print("Setelah reasoning: Reasoner HermiT selesai berjalan.")
```

```python
onto.save(file="BallpointPenOntology_With_Reasoning.owl", format="rdfxml")
print("File berhasil disimpan!")
```
